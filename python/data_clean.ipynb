```python
df_dates = pd.read_excel(r'C:\Data_Analytics_High\farmers_raw_full.xlsx')['registration_date']

def parse_date(value):
    if pd.isna(value):
        return pd.NaT
    value = str(value).strip()
    for fmt in ('%d/%m/%Y', '%Y/%m/%d', '%d-%m-%Y', '%Y-%m-%d'):
        try:
            return pd.to_datetime(value, format=fmt)
        except ValueError:
            continue
    return pd.NaT

df['registration_date'] = df_dates.apply(parse_date)
print(df['registration_date'].isnull().sum())
```

```python
df.to_excel('farmers_cleaned.xlsx', index=False)
```

```python
df = pd.read_excel(r'C:\Data_Analytics_High\operators_raw.xlsx')
df.info()
df = df.drop_duplicates(subset=['operator_id'])
df.loc[df['age'] < 0, 'age'] = np.nan
df['state'] = df['state'].replace({'madhya pradesh':'Madhya Pradesh',' Madhya Pradesh':'Madhya Pradesh','MP':'Madhya Pradesh','M.P.':'Madhya Pradesh','RJ':'Rajasthan','MH':'Maharashtra'})
df['state'] = df['state'].str.title()
df['district'] = df['district'].str.strip().str.title()
df['status'] = df['status'].str.strip().str.title()

df_dates = pd.read_excel(r'C:\Data_Analytics_High\operators_raw.xlsx')['join_date']
df['join_date'] = df_dates.apply(parse_date)

df['phone'] = df['phone'].apply(lambda x: str(int(x)) if pd.notna(x) else np.nan)

print(df.shape)
print(df['state'].unique())
print(df['status'].unique())
print(df['join_date'].isnull().sum())

df.to_excel('operators_cleaned.xlsx', index=False)
```

```python
df = pd.read_excel(r'C:\Data_Analytics_High\machines_raw.xlsx')
df.info()

df = df.drop_duplicates(subset=['machine_id'])

df['machine_type'] = df['machine_type'].str.strip().str.title()
df['district'] = df['district'].str.strip().str.title()
df['status'] = df['status'].str.strip().str.title()

df.loc[df['hourly_rate'] < 0, 'hourly_rate'] = np.nan

operators_clean = pd.read_excel(r'operators_cleaned.xlsx')
df['operator_orphan'] = ~df['operator_id'].isin(operators_clean['operator_id'])

print(df.shape)
print(df['machine_type'].unique())
print(df['status'].unique())
print((df['hourly_rate'] < 0).sum())
print(df['operator_orphan'].sum())

df.to_excel('machines_cleaned.xlsx', index=False)
```

```python
df = pd.read_excel(r'C:\Data_Analytics_High\bookings_raw.xlsx')
df.info()

df = df.drop_duplicates(subset=['booking_id'])

machines_clean = pd.read_excel('machines_cleaned.xlsx')
df['machine_orphan'] = ~df['machine_id'].isin(machines_clean['machine_id'])

df['district'] = df['district'].str.strip().str.title()
df['crop'] = df['crop'].str.strip().str.title()
df['crop'] = df['crop'].replace({'Soyabean': 'Soybean'})
df['status'] = df['status'].str.strip().str.title()

df['acreage_outlier'] = df['acreage'] > 100
df.loc[df['price'] < 0, 'price'] = np.nan

df_dates = pd.read_excel(r'C:\Data_Analytics_High\bookings_raw.xlsx')['booking_date']
df['booking_date'] = df_dates.apply(parse_date)

print(df.shape)
print(df['crop'].unique())
print(df['status'].unique())
print(df['machine_orphan'].sum())
print(df['acreage_outlier'].sum())
print((df['price'] < 0).sum())
print(df['booking_date'].isnull().sum())

df.to_excel('bookings_cleaned.xlsx', index=False)
```
